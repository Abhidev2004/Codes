### Web Scrapping using BeautifulSoup

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.timeanddate.com/weather/india/bangalore"
page = requests.get(url,headers ={"User-Agent":"Mozilla/5.0"})
soup = BeautifulSoup(page.text,"html.parser")

temperature = soup.find("div",class_="h2")
print("Scarpped temperature:",temperature .text.strip() if temperature else "Not Found")
```

### Fetch Current Weather Data

```python
import requests
import json
import pandas as pd
from datetime import datetime
 
API_KEY = "YOUR_API_KEY"
CITY = "Bengaluru"
 
current_url = "https://api.openweathermap.org/data/2.5/weather"
params = {"q": CITY, "appid": API_KEY, "units": "metric"}
 
resp = requests.get(current_url, params=params)
resp.raise_for_status()          # raises an error if the request failed
current_data = resp.json()
 
print(json.dumps(current_data, indent=2))
```

### Fetch 5-Day / 3-Hour Forecast Data

```python
forecast_url = "https://api.openweathermap.org/data/2.5/forecast"
resp2 = requests.get(forecast_url, params=params)
resp2.raise_for_status()
forecast_data = resp2.json()
print("Number of forecast entires:", len(forecast_data["list"]))
```

```python
with open("current_weather.json", "w") as f:
    json.dump(current_data, f, indent=4)
with open("forecast_weather.json", "w") as f:
    json.dump(forecast_data, f, indent=4)

print("JSON files saved successfully.")
```

```python
# Current Weather record
from datetime import datetime, timezone

current_record = {
    "city": current_data["name"],
    "datetime": datetime.fromtimestamp(current_data["dt"], timezone.utc).strftime("%Y-%m-%d %H:%M:%S"),
    "temperature": current_data["main"]["temp"],
    "humidity": current_data["main"]["humidity"],
    "pressure_hpa": current_data["main"]["pressure"],
    "wind_speed_mps": current_data["wind"]["speed"],
    "weather_description": current_data["weather"][0]["description"],
}

pd.DataFrame([current_record]).to_csv("current_weather.csv", index=False)
print("Current weather CSV saved successfully.")
```

```python
# --- Forecast records (every 3-hour slot) ---
forecast_records = []
for entry in forecast_data["list"]:
    forecast_records.append({
        "city": forecast_data["city"]["name"],
        "datetime": entry["dt_txt"],
        "temperature_C": entry["main"]["temp"],
        "humidity_%": entry["main"]["humidity"],
        "pressure_hPa": entry["main"]["pressure"],
        "wind_speed_mps": entry["wind"]["speed"],
        "weather_description": entry["weather"][0]["description"],
    })
 
forecast_df = pd.DataFrame(forecast_records)
forecast_df.to_csv("forecast_weather.csv", index=False)
print(forecast_df.head())
```
